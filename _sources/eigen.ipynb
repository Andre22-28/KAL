# Eigen value dan Eigen Vektor
## Definisi Eigen value dan Eigen Vektor

Dalam aljabar linear, eigenvalue (nilai eigen) dan eigenvector (vektor eigen) adalah konsep yang digunakan untuk memahami bagaimana sebuah matriks memengaruhi suatu vektor. Misalkan terdapat matriks kuadrat 
𝐴 berukuran 𝑛×n. Vektor tak nol v disebut sebagai eigenvector dari matriks 𝐴 jika berlaku persamaan:

$$
A \mathbf{v} = \lambda \mathbf{v}
$$

## Mencari Eigen value dan Eigen vektor
## Contoh soal

$$
A = \begin{bmatrix} 3 & 0 \\ 0 & 3 \end{bmatrix}
$$

$$
B = \begin{bmatrix} -2 & 0 \\ 0 & 4 \end{bmatrix}
$$

## Jawaban

$
\textbf{Soal A}
$

$$
A = \begin{bmatrix} 3 & 0 \\ 0 & 3 \end{bmatrix}
$$

$
\textbf{Langkah 1: Mencari Eigenvalue}
$

$
\textbf{Kita cari nilai \(\lambda\) dari persamaan karakteristik:}
$

$$
\det(A - \lambda I) = 0
$$

$$
\det\left(\begin{bmatrix} 3 - \lambda & 0 \\ 0 & 3 - \lambda \end{bmatrix}\right) = (3 - \lambda)^2 = 0
$$

$$
\Rightarrow \lambda = 3 \text{ (multiplicity 2)}
$$

$
\textbf{{Langkah 2: Mencari Eigenvektor}}
$

$$
(A - 3I)\vec{v} = \mathbf{0}
$$

$$
\Rightarrow \begin{bmatrix} 0 & 0 \\ 0 & 0 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}
$$

$
\textbf{Maka, semua vektor tak nol di \(\mathbb{R}^2\) adalah eigenvektor. Contoh:}
$

$$
\vec{v}_1 = \begin{bmatrix} 1 \\ 0 \end{bmatrix}, \quad \vec{v}_2 = \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$

$
\textbf{Soal B}
$

$
\textbf{Langkah 1: Mencari Eigenvalue}
$

$$
\det(B - \lambda I) = \det\left( \begin{bmatrix} -2 - \lambda & 0 \\ 0 & 4 - \lambda \end{bmatrix} \right) = (-2 - \lambda)(4 - \lambda) = 0
$$

$$
\Rightarrow \lambda_1 = -2, \quad \lambda_2 = 4
$$

$
\textbf{Langkah 2: Mencari Eigenvektor}
$

$$
(B + 2I)\vec{v} = \begin{bmatrix} 0 & 0 \\ 0 & 6 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}
$$

$$
\Rightarrow y = 0, \quad x \text{ bebas}
$$

$$
\Rightarrow \vec{v}_1 = \begin{bmatrix} 1 \\ 0 \end{bmatrix}
$$

$
\textbf{Untuk \(\lambda_2 = 4\):}
$

$$
(B - 4I)\vec{v} = \begin{bmatrix} -6 & 0 \\ 0 & 0 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}
$$

$$
\Rightarrow x = 0, \quad y \text{ bebas}
$$

$$
\Rightarrow \vec{v}_2 = \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$

```python
import numpy as np

# Matriks A
A = np.array([[3, 0],
              [0, 3]])

# Matriks B
B = np.array([[-2, 0],
              [0, 4]])

def print_eigen(name, matrix):
    eigvals, eigvecs = np.linalg.eig(matrix)
    print(f"Eigenvalue dari matriks {name}:")
    print(eigvals)
    print(f"Eigenvektor dari matriks {name}:")
    print(eigvecs)
    print("\n")

# Hitung dan cetak untuk A
print_eigen("A", A)

# Hitung dan cetak untuk B
print_eigen("B", B)
```
