# Definisi Kalisilang

Kalisilang adalah sebuah permainan atau kegiatan teka-teki berbentuk kotak-kotak berbaris dan berkolom yang harus diisi dengan huruf-huruf untuk membentuk kata, berdasarkan petunjuk yang diberikan. Kata-kata tersebut dapat ditulis secara mendatar (horizontal) maupun menurun (vertikal). Kalisilang menguji kemampuan kosakata, logika, serta pengetahuan umum pemainnya.

## Rumus

$$
\vec{u} = \begin{bmatrix} u_1 \\ u_2 \\ u_3 \end{bmatrix}, \quad
\vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ v_3 \end{bmatrix}
$$

$
\textbf{Maka, hasil perkalian silang:}
$

$$
\vec{u} \times \vec{v} =
\begin{bmatrix}
u_2 v_3 - u_3 v_2 \\
u_3 v_1 - u_1 v_3 \\
u_1 v_2 - u_2 v_1
\end{bmatrix}
$$

## Determinan Kovaktor I J K

$
\textbf{Misalkan kita punya matriks:}
$

$$
A = \begin{bmatrix}
i & j & k \\
u_1 & u_2 & u_3 \\
v_1 & v_2 & v_3
\end{bmatrix}
$$

$
\textbf{{Minor} dari elemen \( a_{11} = i \) adalah determinan submatriks dengan menghapus baris 1 dan kolom 1:}
$

$$
M_{11} = \det \begin{bmatrix}
u_2 & u_3 \\
v_2 & v_3
\end{bmatrix} = u_2 v_3 - u_3 v_2
$$

$
\textbf{{Kofaktor} dari elemen \( a_{11} \) adalah:}
$

$$
C_{11} = (-1)^{1+1} M_{11} = +1 \times (u_2 v_3 - u_3 v_2) = u_2 v_3 - u_3 v_2
$$

$
\textbf{Minor dari elemen \( a_{12} = j \):}
$

$$
M_{12} = \det \begin{bmatrix}
u_1 & u_3 \\
v_1 & v_3
\end{bmatrix} = u_1 v_3 - u_3 v_1
$$

$
\textbf{Kofaktor dari elemen \( a_{12}\):}
$

$$
C_{12} = (-1)^{1+2} M_{12} = -1 \times (u_1 v_3 - u_3 v_1) = u_3 v_1 - u_1 v_3
$$
