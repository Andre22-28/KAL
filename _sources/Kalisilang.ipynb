# Definisi Kalisilang  (Cross Product)

Kalisilang adalah sebuah permainan atau kegiatan teka-teki berbentuk kotak-kotak berbaris dan berkolom yang harus diisi dengan huruf-huruf untuk membentuk kata, berdasarkan petunjuk yang diberikan. Kata-kata tersebut dapat ditulis secara mendatar (horizontal) maupun menurun (vertikal). Kalisilang menguji kemampuan kosakata, logika, serta pengetahuan umum pemainnya.

## Rumus

$$
\vec{u} = \begin{bmatrix} u_1 \\ u_2 \\ u_3 \end{bmatrix}, \quad
\vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ v_3 \end{bmatrix}
$$

$
\textbf{Maka, hasil perkalian silang:}
$

$$
\vec{u} \times \vec{v} =
\begin{bmatrix}
u_2 v_3 - u_3 v_2 \\
u_3 v_1 - u_1 v_3 \\
u_1 v_2 - u_2 v_1
\end{bmatrix}
$$

## Determinan Kovaktor I J K

$
\textbf{Misalkan kita punya matriks:}
$

$$
A = \begin{bmatrix}
i & j & k \\
u_1 & u_2 & u_3 \\
v_1 & v_2 & v_3
\end{bmatrix}
$$

$
\textbf{Minor dari elemen \( a_{11} = i \) adalah determinan submatriks dengan menghapus baris 1 dan kolom 1:}
$

$$
M_{11} = \det \begin{bmatrix}
u_2 & u_3 \\
v_2 & v_3
\end{bmatrix} = u_2 v_3 - u_3 v_2
$$

$
\textbf{Kofaktor dari elemen \( a_{11} \) adalah:}
$

$$
C_{11} = (-1)^{1+1} M_{11} = +1 \times (u_2 v_3 - u_3 v_2) = u_2 v_3 - u_3 v_2
$$

$
\textbf{Minor dari elemen \( a_{12} = j \):}
$

$$
M_{12} = \det \begin{bmatrix}
u_1 & u_3 \\
v_1 & v_3
\end{bmatrix} = u_1 v_3 - u_3 v_1
$$

$
\textbf{Kofaktor dari elemen \( a_{12}\):}
$

$$
C_{12} = (-1)^{1+2} M_{12} = -1 \times (u_1 v_3 - u_3 v_1) = u_3 v_1 - u_1 v_3
$$

## TUGAS


### SOAL 1

$
\textbf{Tentukan luas jajaran genjang yang ditentukan oleh vektor }
$

$$
\vec{u} = \begin{bmatrix} 1 \\ 2 \end{bmatrix}, \quad \vec{v} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}
$$

$
\textbf{Peneyelesaian}
$

$
\textbf{Untuk dua vektor dalam ruang 2 dimensi, luas jajaran genjang dapat dihitung dengan nilai mutlak dari determinan matriks 2x2 yang dibentuk oleh kedua vektor, atau menggunakan rumus:
}
$

$$
L = |\vec{u} \times \vec{v}| = |u_1 v_2 - u_2 v_1|
$$

$
\textbf{Substitusi komponen vektor:}
$

$$
\vec{u} = \begin{bmatrix} u_1 \\ u_2 \end{bmatrix} = \begin{bmatrix} 1 \\ 2 \end{bmatrix}, \quad 
\vec{v} = \begin{bmatrix} v_1 \\ v_2 \end{bmatrix} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}
$$

$
\textbf{Maka:}
$

$$
L = |(1)(1) - (2)(2)| = |1 - 4| = |-3| = 3
$$

$
\textbf{Jawaban:}
$

$
\textbf{Luas jajaran genjang adalah:}
$

$$
\boxed{3}
$$

### SOAL 3

$
\textbf{Tentukan luas segitiga dengan titik-titik sudut: }
$

$$
A = (0, 0, 0), \quad B = (1, 3, -1), \quad C = (2, 1, 1)
$$

$
\textbf{Penyelesaian}
$

$
\textbf{Untuk menghitung luas segitiga yang dibentuk oleh tiga titik di ruang 3 dimensi, kita dapat menggunakan rumus:}
$

$$
L = \frac{1}{2} \left\| \vec{AB} \times \vec{AC} \right\|
$$

$
\textbf{Langkah 1: Tentukan vektor} \vec{AB}  dan  \vec{AC} 
$

$$
\vec{AB} = B - A = (1 - 0,\ 3 - 0,\ -1 - 0) = (1, 3, -1)
$$

$$
\vec{AC} = C - A = (2 - 0,\ 1 - 0,\ 1 - 0) = (2, 1, 1)
$$

$
\textbf{Langkah 2: Hitung}  \vec{AB} \times \vec{AC}
$

$$
\vec{AB} \times \vec{AC}
$$


$$
\vec{AB} \times \vec{AC} = 
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
1 & 3 & -1 \\
2 & 1 & 1 \\
\end{vmatrix}
= \hat{i}(3 \cdot 1 - (-1) \cdot 1) - \hat{j}(1 \cdot 1 - (-1) \cdot 2) + \hat{k}(1 \cdot 1 - 3 \cdot 2)
$$

$$
= \hat{i}(3 + 1) - \hat{j}(1 + 2) + \hat{k}(1 - 6)
= \hat{i}(4) - \hat{j}(3) - \hat{k}(5)
$$

$
\textbf{Langkah 3: Hitung panjang vektor hasil cross product}
$

$$
\left\| \vec{AB} \times \vec{AC} \right\| = \sqrt{4^2 + (-3)^2 + (-5)^2} = \sqrt{16 + 9 + 25} = \sqrt{50}
$$

$
\textbf{Langkah 4: Hitung luas segitiga}
$

$$
L = \frac{1}{2} \sqrt{50} = \frac{\sqrt{50}}{2}
$$

$
\textbf{Jawaban:}
$

$
\textbf{Luas segitiga adalah:}
$

$$
\boxed{\frac{\sqrt{50}}{2}}
$$

### SOAL 3


$
\textbf{Tentukan luas segitiga dengan titik-titik sudut: }
$

$$
A = (5, 2, -1), \quad B = (3, 6, 2), \quad C = (1, 0, 4)
$$

$
\textbf{Penyelesaian}
$

$
\textbf{Untuk menghitung luas segitiga dari tiga titik di ruang 3 dimensi, gunakan rumus:}
$

$$
L = \frac{1}{2} \left\| \vec{AB} \times \vec{AC} \right\|
$$

$
\textbf{Langkah 1: Hitung vektor} \vec{AB}  dan \vec{AC} 
$

$$
\vec{AB} = B - A = (3 - 5,\ 6 - 2,\ 2 - (-1)) = (-2,\ 4,\ 3)
$$

$$
\vec{AC} = C - A = (1 - 5,\ 0 - 2,\ 4 - (-1)) = (-4,\ -2,\ 5)
$$

$
\textbf{Langkah 2: Hitung} \vec{AB} \times \vec{AC} 
$

$$
\vec{AB} \times \vec{AC} = 
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
-2 & 4 & 3 \\
-4 & -2 & 5 \\
\end{vmatrix}
= \hat{i}(4 \cdot 5 - 3 \cdot (-2)) - \hat{j}(-2 \cdot 5 - 3 \cdot (-4)) + \hat{k}(-2 \cdot (-2) - 4 \cdot (-4))
$$

$$
= \hat{i}(20 + 6) - \hat{j}(-10 + 12) + \hat{k}(4 + 16)
= \hat{i}(26) - \hat{j}(2) + \hat{k}(20)
$$

$
\textbf{Langkah 3: Hitung panjang vektor hasil cross product}
$

$$
\left\| \vec{AB} \times \vec{AC} \right\| = \sqrt{26^2 + 2^2 + 20^2} = \sqrt{676 + 4 + 400} = \sqrt{1080}
$$

$
\textbf{Langkah 4: Hitung luas segitiga}
$

$$
L = \frac{1}{2} \sqrt{1080}
$$

$
\textbf{Jawaban}
$

$
\text{Luas segitiga adalah:}
$

$$
\boxed{\frac{\sqrt{1080}}{2}}
$$

$

$$
\vec{AB} \times \vec{AC} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\

\end{vmatrix}

$$
