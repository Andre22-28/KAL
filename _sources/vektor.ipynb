## Definisi Perkalian Titik (Dot Product) Vektor
Perkalian titik (dot product) adalah operasi matematika antara dua vektor yang menghasilkan sebuah skalar (bilangan real). Dot product menunjukkan hubungan arah antara dua vektor dan digunakan untuk menghitung sudut di antara keduanya, serta menentukan apakah kedua vektor searah, tegak lurus, atau berlawanan arah.

Secara matematis, untuk dua vektor 𝑎 dan 𝑏, dot product ditulis sebagai:


$$
\vec{a} \cdot \vec{b}
$$

```python
def dot_product(a, b):
    if len(a) != len(b):
        raise ValueError("Panjang kedua vektor harus sama.")
    return sum(x * y for x, y in zip(a, b))

# Contoh penggunaan
vektor1 = [3, 4]
vektor2 = [2, -1]

hasil = dot_product(vektor1, vektor2)
print("Dot product:", hasil)
```

## Definisi Vektor Satuan
Vektor satuan adalah vektor yang memiliki besar (magnitudo) sama dengan 1 satuan, tetapi tetap menunjukkan arah tertentu di ruang.

Vektor ini digunakan untuk menyatakan arah saja, tanpa memperhatikan panjang aslinya. Biasanya dilambangkan dengan huruf bertopi, misalnya:


$$
\hat{a} = \frac{\vec{a}}{|\vec{a}|}
$$


```python
import math

def magnitude(vector):
    return math.sqrt(sum(comp**2 for comp in vector))

def unit_vector(vector):
    mag = magnitude(vector)
    if mag == 0:
        raise ValueError("Vektor nol tidak memiliki arah sehingga tidak bisa dibuat vektor satuan.")
    return [comp / mag for comp in vector]

# Contoh penggunaan
vektor = [3, 4]
vektor_satuan = unit_vector(vektor)
magnitudo_satuan = magnitude(vektor_satuan)

print("Vektor asal:", vektor)
print("Vektor satuan:", vektor_satuan)
print("Magnitudo vektor satuan:", magnitudo_satuan)
```
