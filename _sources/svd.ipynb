# Singular Value Decomposition

## Apa itu SVD

**Singular Value Decomposition (SVD)** adalah salah satu teknik dalam **aljabar linear** yang sangat penting, terutama dalam **analisis data**, **kompresi**, dan **machine learning**. Secara sederhana, SVD memecah sebuah matriks menjadi tiga komponen utama yang lebih sederhana untuk dianalisis atau dimanipulasi.

Jika ada sebuah matriks $A$ berukuran $m \times n$, maka **SVD** menuliskan matriks tersebut sebagai:

$$
A = U \Sigma V^T
$$


## **Kegunaan Singular Value Decomposition (SVD)**

1. 🔹 **Reduksi Dimensi (Dimensionality Reduction)**

   * Mengurangi jumlah fitur dalam data, tanpa menghilangkan informasi penting.
   * Contoh: PCA (Principal Component Analysis).

2. 🔹 **Kompresi Data dan Gambar**

   * Mengecilkan ukuran file (gambar, video, dll) dengan tetap mempertahankan kualitas utama.
   * Sering digunakan dalam pengolahan citra.

3. 🔹 **Rekomendasi Sistem (Recommendation System)**

   * Memprediksi preferensi pengguna berdasarkan data interaksi sebelumnya.
   * Contoh: Netflix, YouTube, dan Tokopedia pakai ini.

4. 🔹 **Pembersihan Data / Penghilangan Noise**

   * Menghilangkan informasi yang dianggap “gangguan” atau noise dalam data numerik.

5. 🔹 **Pencarian dan Analisis Teks (LSA/LSI)**

   * Menemukan hubungan tersembunyi antara kata dan dokumen.
   * Digunakan dalam mesin pencari dan pengelompokan dokumen.

6. 🔹 **Solusi Sistem Persamaan Linier**

   * Membantu menyelesaikan sistem linier yang tidak memiliki solusi unik atau terlalu banyak variabel.

7. 🔹 **Analisis Struktur dan Korelasi Data**

   * Menemukan pola dan hubungan dalam dataset besar (misalnya data pasar, keuangan, dll).


###  Singkatnya:

 **SVD berguna untuk menyederhanakan, menganalisis, dan mengekstrak informasi penting dari data dalam bentuk matriks.**




## **langkah-langkah mencari SVD (Singular Value Decomposition)** secara manual dari suatu matriks:


#### CONTOH: Misalnya kita punya matriks

$$
A = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
$$

Kita ingin menemukan $A = U \Sigma V^T$

#### **Langkah 1: Hitung $A^T A$**

$$
A^T = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
\Rightarrow
A^T A = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
\begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
=
\begin{bmatrix}
10 & 6 \\
6 & 10
\end{bmatrix}
$$


#### **Langkah 2: Cari eigenvalue dan eigenvector dari $A^T A$**

$$
\text{Det}(A^T A - \lambda I) = 0
$$

$$
\begin{vmatrix}
10 - \lambda & 6 \\
6 & 10 - \lambda
\end{vmatrix}
= (10 - \lambda)^2 - 36 = 0
\Rightarrow (10 - \lambda)^2 = 36
\Rightarrow \lambda = 4, 16
$$

Maka:

* Singular values:

$$
\sigma_1 = \sqrt{16} = 4, \quad \sigma_2 = \sqrt{4} = 2
$$

#### **Langkah 3: Cari eigenvector dari $A^T A$**

Untuk $\lambda = 16$:

$$
(A^T A - 16I) v = 0
\Rightarrow
\begin{bmatrix}
-6 & 6 \\
6 & -6
\end{bmatrix}
v = 0
\Rightarrow v_1 = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$

Untuk $\lambda = 4$:

$$
(A^T A - 4I) v = 0
\Rightarrow
\begin{bmatrix}
6 & 6 \\
6 & 6
\end{bmatrix}
v = 0
\Rightarrow v_2 = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ -1 \end{bmatrix}
$$


#### **Langkah 4: Bangun matriks $V$**

$$
V = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}
$$


#### **Langkah 5: Hitung $U$ menggunakan rumus $u_i = \frac{1}{\sigma_i} A v_i$**

Untuk $u_1$:

$$
u_1 = \frac{1}{4} A v_1 = \frac{1}{4} A \left( \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix} \right)
= \frac{1}{4\sqrt{2}} \begin{bmatrix} 3+1 \\ 1+3 \end{bmatrix} = \frac{1}{4\sqrt{2}} \begin{bmatrix} 4 \\ 4 \end{bmatrix} = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$

Untuk $u_2$:

$$
u_2 = \frac{1}{2} A v_2 = \frac{1}{2} A \left( \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ -1 \end{bmatrix} \right)
= \frac{1}{2\sqrt{2}} \begin{bmatrix} 3 - 1 \\ 1 - 3 \end{bmatrix}
= \frac{1}{2\sqrt{2}} \begin{bmatrix} 2 \\ -2 \end{bmatrix}
= \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ -1 \end{bmatrix}
$$


####  **Langkah 6: Susun hasil akhir**

$$
U = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix},
\quad
\Sigma = \begin{bmatrix}
4 & 0 \\
0 & 2
\end{bmatrix},
\quad
V^T = V^\top = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}
$$

#### **Hasil akhir:**

$$
A = U \Sigma V^T
$$

## untuk matriks:

$$
A = \begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}
$$

#### **Langkah 1: Hitung $A^T A$**

$$
A^T A = \begin{bmatrix}
3 & 2 & 4 & 1 \\
7 & 5 & 3 & 1
\end{bmatrix}
\cdot
\begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}
=
\begin{bmatrix}
30 & 44 \\
44 & 84
\end{bmatrix}
$$

#### **Langkah 2: Hitung Eigenvalue dan Eigenvector dari $A^T A$**

* Eigenvalue (λ):

  $$
  \lambda_1 = 108.62,\quad \lambda_2 = 5.38
  $$

* Vektor eigen (kolom-kolom matriks $V$):

$$
V =
\begin{bmatrix}
-0.4884 & -0.8726 \\
-0.8726 & 0.4884
\end{bmatrix}
$$


#### **Langkah 3: Hitung nilai singular (akar dari eigenvalue)**

$$
\sigma_1 = \sqrt{108.62} \approx 10.42,\quad
\sigma_2 = \sqrt{5.38} \approx 2.32
$$


#### **Langkah 4: Bentuk matriks $\Sigma$ (ukuran 4×2 karena A = 4×2)**

$$
\Sigma =
\begin{bmatrix}
10.42 & 0 \\
0 & 2.32 \\
0 & 0 \\
0 & 0
\end{bmatrix}
$$


#### **Langkah 5: Hitung matriks $U = A V \Sigma^{-1}$**

$$
U \approx
\begin{bmatrix}
-0.727 & 0.345 \\
-0.512 & 0.300 \\
-0.439 & -0.874 \\
-0.131 & -0.166
\end{bmatrix}
$$

#### **Hasil Akhir: Dekomposisi SVD**

$$
A = U \Sigma V^T
$$

Dengan:

* $U$: matriks ortonormal $4 \times 2$
* $\Sigma$: matriks diagonal $4 \times 2$
* $V^T$: transpose dari matriks ortonormal $2 \times 2$
